```python
from ultralytics import YOLO

# Load the YOLOv8 small model
model = YOLO('yolov8s.pt')

# Train the model
model.train(
    data='/content/dataset/data.yaml',
    epochs=50,
    imgsz=640,
    batch=16,
    project='/content/drive/MyDrive/traffic_model'
)
```

```python
from ultralytics import YOLO
from IPython.display import Image, display

# Load your newly trained model from your Drive
model = YOLO('/content/drive/MyDrive/traffic_model/train/weights/best.pt')

# Run inference on a test image (replace with an actual image name from your test folder)
results = model.predict(source='/content/dataset/test/images/1-Gap-in-Median_117_png.rf.234c03c862f5252ccca7a24619ff5452.jpg', save=True)

# Display the result directly in Colab
# Display the result directly in Colab
display(Image(filename=results[0].save_dir + '/' + '1-Gap-in-Median_117_png.rf.234c03c862f5252ccca7a24619ff5452.jpg'))
```
